```python
# Remove duplicates
for name in df_dict:
    initial_count = len(df_dict[name])

    # drop_duplicates() removes identical rows
    # inplace=True modifies the DataFrame directly in the dictionary
    df_dict[name].drop_duplicates(inplace=True)

    final_count = len(df_dict[name])
    removed = initial_count - final_count

    print(f"Removed {removed} duplicates from {{{name}}}. New total: {final_count}")

# 3. Re-assign back to original variables if needed
df_enrol = df_dict["df_enrol"]
df_demographic = df_dict["df_demographic"]
df_biometric = df_dict["df_biometric"]
```

```python
# Function to make a unique name to a state.
def clean_state_names(df, column='state'):
    # Convert to string, lowercase, and strip whitespace
    df[column] = df[column].astype(str).str.lower().str.strip()

    # Create a mapping for variations
    state_mapping = {
        'west bengal': 'West Bengal',
        'west  bengal': 'West Bengal',
        'west bangal': 'West Bengal',
        'westbengal': 'West Bengal',
        'west bengal': 'West Bengal',
        'jammu and kashmir': 'Jammu and Kashmir',
        'jammu & kashmir': 'Jammu and Kashmir',
        'odisha': 'Odisha',
        'orissa': 'Odisha',
        'andhra pradesh': 'Andhra Pradesh',
        'the dadra and nagar haveli and daman and diu': 'Dadra and Nagar Haveli and Daman and Diu',
        'dadra and nagar haveli and daman and diu': 'Dadra and Nagar Haveli and Daman and Diu',
        'dadra & nagar haveli': 'Dadra and Nagar Haveli and Daman and Diu',
        'dadra and nagar haveli': 'Dadra and Nagar Haveli and Daman and Diu',
        'daman and diu': 'Dadra and Nagar Haveli and Daman and Diu',
        'daman & diu': 'Dadra and Nagar Haveli and Daman and Diu',
        'pondicherry': 'Puducherry',
        'puducherry': 'Puducherry',
        'andaman and nicobar islands': 'Andaman and Nicobar Islands',
        'andaman & nicobar islands': 'Andaman and Nicobar Islands'
    }

    # Apply mapping, then title case for consistent capitalization of others
    df[column] = df[column].replace(state_mapping).apply(lambda x: x.title() if x not in state_mapping.values() else x)

    return df

# Apply cleaning to df_enrol
df_enrol = clean_state_names(df_enrol, 'state')

# Remove rows where state is '100000'
df_enrol = df_enrol[df_enrol['state'] != '100000']

# Display unique states and shape to verify
print("Unique states after cleaning:")
print(df_enrol['state'].unique())
print(f"\nShape of df_enrol after cleaning: {df_enrol.shape}")
```

```python
def aggregate_data(df):

    aggregation_logic = {
        'age_0_5': 'sum',
        'age_5_17': 'sum',
        'age_18_greater': 'sum',
        'state': 'first',    # Keeps the state name without trying to add it
        'month': 'first',    # Keeps the month without trying to add it
        'pincode': 'first',    # Keeps the pincode without trying to add it
        'district': 'first'   # Keeps the region name
    }

    # 2. Group and aggregate
    result = df.groupby(['month', 'pincode'], as_index=False).agg(aggregation_logic)

    return result
```

```python

district_stats = df_enrol.groupby(['state', 'district'])[['age_0_5', 'age_5_17']].sum().reset_index()

# Filter out tiny districts to keep the chart clean
# We only keep districts with at least 100 total child enrollments
district_stats['total_kids'] = district_stats['age_0_5'] + district_stats['age_5_17']
district_stats = district_stats[district_stats['total_kids'] > 100]

# STEP 1: CALCULATE CORRELATION
corr = district_stats['age_0_5'].corr(district_stats['age_5_17'])

# STEP 2: PLOT THE RELATIONSHIP
fig = px.scatter(
    district_stats,
    x='age_0_5',
    y='age_5_17',
    color='state',          # Color dots by State to see regional clusters
    hover_name='district',  # Show District Name on hover
    size='total_kids',

    # Log scale is CRITICAL here because Indian districts vary wildly in size

    log_x=True,
    log_y=True,

    trendline="ols",        # Adds the Regression Line (General Trend)
    trendline_scope="overall", # One single trendline for all India
    trendline_color_override="black",

    title=f"District Dynamics: Newborns (0-5) vs. School Age (5-17) [Correlation: {corr:.2f}]",
    labels={
        "age_0_5": "Newborn Enrollments (0-5) [Log Scale]",
        "age_5_17": "School-Age Enrollments (5-17) [Log Scale]",
        "state": "State"
    },
    opacity=0.7
)

# STEP 4: ADD DIAGNOSTIC ZONES
# We add a diagonal line (y=x).
# Districts ABOVE this line are "School-Heavy".
# Districts BELOW this line are "Birth-Heavy".
fig.add_shape(
    type="line", line=dict(dash="dash", color="red", width=1),
    x0=1, y0=1, x1=100000, y1=100000 # Spanning the log range
)

fig.update_layout(height=700, plot_bgcolor='white')
fig.show()
```

```python
# We aggregate the total counts for the three age buckets by Gender
# Note: df_enrol usually separates age cols and gender cols.
# If your data doesn't have "Male 0-5" specifically, we use the broad Gender totals
# relative to Age totals to estimate the split, OR we plot Age Groups directly.

# METHOD: Since standard data often gives Age OR Gender but not "Male-Age-0-5",
# We will plot the 3 Age Groups (the "Pyramid" levels) for the total population.

# Aggregate total enrollment by Age Group
age_counts = df_enrol[['age_0_5', 'age_5_17', 'age_18_greater']].sum().reset_index()
age_counts.columns = ['Age_Group', 'Count']

# Rename for clarity
label_map = {
    'age_0_5': '0-5 (Newborns)',
    'age_5_17': '5-17 (School Age)',
    'age_18_greater': '18+ (Adults)'
}
age_counts['Age_Label'] = age_counts['Age_Group'].map(label_map)

# Calculate Percentage to see the "Structure" clearly
total_population = age_counts['Count'].sum()
age_counts['Percentage'] = (age_counts['Count'] / total_population) * 100

# DEFINE THE BENCHMARK (Census 2011 / Projected 2025)
# Approximate Indian Demographics (Benchmark):
# 0-5: ~10%
# 5-18: ~25%
# 18+: ~65%
# We create a dataframe for this comparison
benchmarks = [10.0, 25.0, 65.0]
age_counts['Benchmark'] = benchmarks

# Calculate "Divergence" (Difference between Reality and Benchmark)
age_counts['Divergence'] = age_counts['Percentage'] - age_counts['Benchmark']


fig = go.Figure()

# Plot 1: The Reality (Bars)
fig.add_trace(go.Bar(
    y=age_counts['Age_Label'],
    x=age_counts['Percentage'],
    name='Actual Enrollment (%)',
    orientation='h',
    marker=dict(color='#1f77b4', line=dict(color='black', width=1))
))

# Plot 2: The Benchmark (Dotted Line / Markers)
fig.add_trace(go.Scatter(
    y=age_counts['Age_Label'],
    x=age_counts['Benchmark'],
    name='Ideal Census Baseline (%)',
    mode='markers+lines',
    marker=dict(color='red', size=15, symbol='line-ns-open'),
    line=dict(color='red', dash='dot')
))

#  FORMATTING
fig.update_layout(
    title="Population Pyramid Divergence: Actual vs. Census Benchmark",
    xaxis_title="Share of Total Enrollment (%)",
    yaxis_title="Age Bucket",
    height=500,
    plot_bgcolor='white',
    legend=dict(x=0.6, y=0.1),
    xaxis=dict(showgrid=True, gridcolor='lightgray')
)

# Add Annotations for the Divergence
for index, row in age_counts.iterrows():
    diff = row['Divergence']
    color = "green" if diff >= -2 else "red" # Green if close/above, Red if huge gap
    label = f"{diff:+.1f}% Gap"

    fig.add_annotation(
        x=row['Percentage'] + 2,
        y=row['Age_Label'],
        text=label,
        showarrow=False,
        font=dict(color=color, size=12)
    )

fig.show()
```

```python
# Function to make a unique name to a state, updated with city-to-state mappings and other corrections.
def clean_state_names(df, column='state'):
    # Convert to string, lowercase, and strip whitespace
    df[column] = df[column].astype(str).str.lower().str.strip()

    # Create a mapping for variations and city names to state names
    state_mapping = {
        'west bengal': 'West Bengal',
        'west  bengal': 'West Bengal',
        'west bangal': 'West Bengal',
        'westbengal': 'West Bengal',
        'west bengli': 'West Bengal', # Corrected misspelling
        'jammu and kashmir': 'Jammu and Kashmir',
        'jammu & kashmir': 'Jammu and Kashmir',
        'odisha': 'Odisha',
        'orissa': 'Odisha',
        'andhra pradesh': 'Andhra Pradesh',
        'the dadra and nagar haveli and daman and diu': 'Dadra and Nagar Haveli and Daman and Diu',
        'dadra and nagar haveli and daman and diu': 'Dadra and Nagar Haveli and Daman and Diu',
        'dadra & nagar haveli': 'Dadra and Nagar Haveli and Daman and Diu',
        'dadra and nagar haveli': 'Dadra and Nagar Haveli and Daman and Diu',
        'daman and diu': 'Dadra and Nagar Haveli and Daman and Diu',
        'daman & diu': 'Dadra and Nagar Haveli and Daman and Diu',
        'pondicherry': 'Puducherry',
        'puducherry': 'Puducherry',
        'andaman and nicobar islands': 'Andaman and Nicobar Islands',
        'andaman & nicobar islands': 'Andaman and Nicobar Islands',
        'chhatisgarh': 'Chhattisgarh', # Corrected misspelling
        'uttaranchal': 'Uttarakhand',  # Corrected misspelling

        # City/locality names to state names
        'darbhanga': 'Bihar',
        'puttenahalli': 'Karnataka',
        'balanagar': 'Telangana',
        'jaipur': 'Rajasthan',
        'madanapalle': 'Andhra Pradesh',
        'nagpur': 'Maharashtra',
        'raja annamalai puram': 'Tamil Nadu'
    }

    # Apply mapping, then title case for consistent capitalization of others
    df[column] = df[column].replace(state_mapping).apply(lambda x: x.title() if x not in state_mapping.values() else x)

    return df

# Apply the updated cleaning function to df_demographic
df_demographic = clean_state_names(df_demographic, 'state')

# Remove rows where state is '100000'
df_demographic = df_demographic[df_demographic['state'] != '100000']

print("Unique states after cleaning:")
print(df_demographic['state'].unique())
print(f"\nShape of df_demographic after cleaning: {df_demographic.shape}")
```

```python
# Group total biometrics by month for each age group
df_bio_age_trend = df_bio.groupby('month')[['bio_age_5_17', 'bio_age_17_']].sum().reset_index()


df_bio_age_trend_melted = df_bio_age_trend.melt(id_vars=['month'],
                                                    var_name='age_group',
                                                    value_name='total_biometric')

# Map age_group names for better readability on the plot
age_group_mapping = {
    'bio_age_5_17': 'Age 5-17',
    'bio_age_17_': 'Age 17+'
}
df_bio_age_trend_melted['age_group'] = df_bio_age_trend_melted['age_group'].map(age_group_mapping)

# Create the line plot
fig = px.line(df_bio_age_trend_melted,
              x='month',
              y='total_biometric',
              color='age_group',
              title='Trend of Total Biometrics by Age Group Across Months',
              labels={'month': 'Month Number', 'total_biometric': 'Total Biometrics', 'age_group': 'Age Group'},
              hover_data={'month': True, 'total_biometric': ':.0f', 'age_group': True},
              markers=True
             )

fig.update_layout(
    xaxis_title='Month Number',
    yaxis_title='Total Biometrics',
    hovermode='x unified'
)

fig.show()
```

```python
df_bio_by_district = df_bio.groupby('district')[['bio_age_5_17', 'bio_age_17_']].sum().reset_index()
df_bio_by_district['total_biometric'] = df_bio_by_district['bio_age_5_17'] + df_bio_by_district['bio_age_17_']

print("Total bioment per district calculated and stored in 'df_bio_by_district'.")
df_bio_by_district.head()
```

```python
df_bio_by_district['bio_age_17_'] = df_bio_by_district.apply(lambda row: row['bio_age_17_'] / row['total_biometric'] if row['total_biometric'] != 0 else 0, axis=1)
print("Calculated 'ratio_17_plus' column in df_bio_by_district.")
df_bio_by_district.head()
```

```python
df_top_20_districts = df_bio_by_district.sort_values(by='bio_age_17_', ascending=False).head(20)
print("Top 20 districts by 'bio_age_17_' ratio calculated and stored in 'df_top_20_districts'.")
df_top_20_districts.head()
```
